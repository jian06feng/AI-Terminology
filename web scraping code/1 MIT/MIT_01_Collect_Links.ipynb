# MIT News: collect AI-tagged article links

Walks the paginated topic listing `news.mit.edu/topic/artificial-intelligence2` (up to 100 pages) and stores each article's title and URL.

Output: `./Data/MIT Articles Links.csv`


```python
import random
from pathlib import Path
from time import sleep

import pandas as pd
import requests
from bs4 import BeautifulSoup
from tqdm.auto import tqdm

BASE_URL = "https://news.mit.edu"
PAGE_URL = f"{BASE_URL}/topic/artificial-intelligence2?page="
DATA_DIR = Path("./Data")
OUTPUT_PATH = DATA_DIR / "MIT Articles Links.csv"
MAX_PAGES = 100

DATA_DIR.mkdir(parents=True, exist_ok=True)
```

```python
article_links = {}

for page in tqdm(range(MAX_PAGES)):
    response = requests.get(f"{PAGE_URL}{page}", timeout=30)
    soup = BeautifulSoup(response.text, "html.parser")

    items = soup.find_all("h3", {"class": "term-page--news-article--item--title"})
    if not items:
        break

    for article in items:
        heading = article.find("span").get_text()
        article_links[heading] = BASE_URL + article.find("a").get("href")

    sleep(random.choice([0.3, 0.5, 0.7]))

article_links_df = pd.DataFrame(article_links.items(), columns=["Article", "Link"])
print(f"Collected {len(article_links_df)} article links.")
article_links_df.head()
```

```python
article_links_df.to_csv(OUTPUT_PATH, index=False)
print(f"Saved to: {OUTPUT_PATH}")
```
